## What is the total amount each customer spent on zomato ?

### SELECT a.userid, a.product_id,b.price FROM sales a INNER JOIN product b ON a.product_id=b.product_id;



SELECT a.userid,sum(b.price) total_amt_spent FROM sales a INNER JOIN product b ON a.product_id=b.product_id
GROUP BY a.userid;

### Explanation of the SQL Code:
This query calculates the **total amount of money each customer spent** on Zomato.

#### How it works step-by-step:
1. **`INNER JOIN product b ON a.product_id = b.product_id`:**
   - Combines the `sales` table (`a`) with the `product` table (`b`) using the shared column `product_id`. This allows us to access the `price` of each item sold.
2. **`GROUP BY a.userid`:**
   - Groups the combined sales data by each unique customer (`userid`).
3. **`SUM(b.price) total_amt_spent`:**
   - Adds up the prices of all the items purchased by each respective customer and labels the resulting column as `total_amt_spent`.

## How many days has each customer visited zomato ?

###
SELECT userid, COUNT(DISTINCT created_date) distinct_days FROM sales
GROUP BY userid;

### Explanation of the SQL Code:
This query finds **how many unique days each customer visited** Zomato.

#### How it works step-by-step:
1. **`COUNT(DISTINCT created_date) distinct_days`:**
   - Identifies and counts only the unique, non-repeating dates (`created_date`) in the database for each customer. This ensures that if a customer makes multiple purchases on the same day, that day is counted only once.
2. **`GROUP BY userid`:**
   - Groups the calculations so that we get a separate count of visited days for each individual customer (`userid`).

## What is the most purchased item on the menu & how many times it was purchased by all customers ?

###
SELECT product_id, COUNT(product_id) FROM sales
GROUP BY product_id
ORDER BY COUNT(product_id) DESC;

SELECT product_id, COUNT(product_id) cnt FROM sales
GROUP BY product_id
ORDER BY COUNT(product_id) DESC LIMIT 1;

### Explanation of the SQL Code:
This query finds the **single most purchased item** on the menu.

#### How it works step-by-step:
1. **`GROUP BY product_id`:**
   - Groups all transaction rows in the `sales` table by their `product_id`.
2. **`COUNT(product_id) cnt`:**
   - Counts how many times each product was purchased.
3. **`ORDER BY COUNT(product_id) DESC`:**
   - Sorts the results from the highest purchase count to the lowest.
4. **`LIMIT 1`:**
   - Keeps only the top record, which represents the most popular product on the menu.

###
SELECT userid, COUNT(product_id) cnt FROM sales
WHERE product_id=(SELECT product_id FROM sales
GROUP BY product_id
ORDER BY COUNT(product_id) DESC LIMIT 1)
GROUP BY userid;

### Explanation of the SQL Code:
This query determines **how many times each individual user purchased the most popular item** on the menu.

#### How it works step-by-step:
1. **Subquery (`SELECT product_id FROM sales GROUP BY product_id ORDER BY COUNT(product_id) DESC LIMIT 1`):**
   - Groups all records in the `sales` table by `product_id`.
   - Counts how many times each product was purchased.
   - Sorts them in descending order and limits the result to `1` to find the single most purchased `product_id` (which is product `2`).

2. **Outer Query `WHERE` Clause:**
   - Filters the `sales` table to only look at transactions involving this specific, most-purchased product ID.

3. **Outer Query `GROUP BY userid`:**
   - Groups the filtered sales records by `userid` and counts (`COUNT(product_id)`) how many times each customer bought that specific product.

## Which item was first purchased by customer after they become a member ?

###
SELECT a.userid, a.created_date, a.product_id, b.gold_signup_date
FROM sales a INNER JOIN goldusers_signup b on a.userid=b.userid;

### Explanation of the SQL Code:
This query is the initial step to find the **first item purchased by a customer after signing up for the gold membership**.

#### How it works step-by-step:
1. **`INNER JOIN goldusers_signup b ON a.userid = b.userid`:**
   - Joins the `sales` table (`a`) with the `goldusers_signup` table (`b`) on `userid` so we can compare each transaction's `created_date` directly with the user's `gold_signup_date`.
2. **`SELECT a.userid, a.created_date, a.product_id, b.gold_signup_date`:**
   - Displays the customer, when they bought a product, what they bought, and when they signed up for the gold membership, laying the groundwork for filtering out transactions before the membership date.
